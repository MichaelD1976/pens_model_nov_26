```python
import numpy as np
df['country'] = np.select(
    [
        df['competition_id'].isin(eng1_ids),
        df['competition_id'].isin(fra1_ids),
        df['competition_id'].isin(spa1_ids),
        df['competition_id'].isin(ger1_ids),
        df['competition_id'].isin(ita1_ids),
    ],
    ['eng1', 'fra1', 'spa1', 'ger1', 'ita1'],
    default='other'
)
```

```python
dfOU = pd.read_csv('../data/raw/OverUnder_Exp.csv')
```

```python
dfOU.head()
```

```python
df['o_prob'] = round(df['odds_ft_over25'] / ((df['odds_ft_over25'] + df['odds_ft_under25'])), 2)
df['u_prob'] = round(df['odds_ft_under25'] / ((df['odds_ft_over25'] + df['odds_ft_under25'])), 2)
df['2_5_margin'] = df['o_prob']  + df['u_prob']
df['o2_5_true'] = df['o_prob']  * df['2_5_margin']
df['u2_5_true'] = df['u_prob']  * df['2_5_margin']
```

```python
dfOU = dfOU.drop(['Exp1', 'Under', 'Over', 'Un2.5_%'], axis=1)
dfOU.drop_duplicates(subset='Ov2.5_%', inplace=True)
# rename so merge possible
dfOU.rename(columns = {'Ov2.5_%': 'o2_5_true'}, inplace=True)
df_new = pd.merge(left=df, right=dfOU, on='o2_5_true')
df_new.rename(columns={'Exp':'G_Exp'}, inplace=True)
```

```python
df_new.head()
```

```python
df=df_new.copy()
```

```python
df['prob_h'] = round(df['odds_ft_1'] / (df['odds_ft_1'] + df['odds_ft_x'] + df['odds_ft_2']), 2)
df['prob_a'] = round(df['odds_ft_2'] / (df['odds_ft_1'] + df['odds_ft_x'] + df['odds_ft_2']), 2)
df['prob_x'] = round(df['odds_ft_x'] / (df['odds_ft_1'] + df['odds_ft_x'] + df['odds_ft_2']), 2)
df['margin_1x2'] = df['prob_h'] + df['prob_a'] + df['prob_x']
df['h_true'] = df['prob_h'] / df['margin_1x2']
df['a_true'] = df['prob_a'] / df['margin_1x2']
df['x_true'] = df['prob_x'] / df['margin_1x2']

df['HG_Exp'] = round((df['G_Exp']) * (df['h_true'] + 0.5 * df['x_true']), 2)
df['AG_Exp'] = round((df['G_Exp']) * (df['a_true'] + 0.5 * df['x_true']), 2)
```

```python
df = df[df['pens_recorded'] ==1]
len(df)
```

```python
df = df[['HG_Exp', 'AG_Exp', 'team_a_penalties_won', 'team_b_penalties_won', 'country']]
```

```python
df.sample(10)
```

```python
# column for if ANY pen was won - doesnt distinguish between 1 or 2 won
df['h_got_any_penalty'] = (df['team_a_penalties_won'] > 0).copy().astype(int)
df['a_got_any_penalty'] = (df['team_b_penalties_won'] > 0).copy().astype(int)
```

```python
df.head()
```

```python
df['any_match_pen'] = ((df['h_got_any_penalty'] == 1) | (df['a_got_any_penalty'] == 1)).astype(int)
```

```python
df.head()
```

```python
print('Shape before cleaning:', df.shape)
print('\nMissing valuesbefore cleaning:')
print(df.isna().sum())

df=df.dropna()
df = df.reset_index(drop=True)

print('\nShape after cleaning:', df.shape)
```
